# PANDORA data exploration

Run the cells in order after `00_preprocess.ipynb`. This notebook reads the main dataset used by the baselines and uses the preprocessing report to summarize comments removed from personality subreddits.

```python
from pathlib import Path

import pandas as pd
from IPython.display import display

DATA = Path("../data/preprocessed")  # From the project's src folder
COMMENTS = DATA / "main/comments.csv"
PROFILES = DATA / "main/users.csv"
REMOVED_SUBREDDITS = DATA / "removed_by_subreddit.csv"

TRAITS = ["agreeableness", "openness", "conscientiousness",
          "extraversion", "neuroticism"]
```

## 1. Load the data and check its size

```python
comments = pd.read_csv(COMMENTS, keep_default_na=False)
profiles = pd.read_csv(PROFILES)

print(f"Comments: {len(comments):,}")
print(f"Users with Big Five scores: {len(profiles):,}")
print(f"Users with comments: {comments['author'].nunique():,}")
```

## 2. Check the number of comments per user

```python
comments_per_user = comments.groupby("author").size()
display(comments_per_user.describe())
```

## 3. Check comment length

```python
word_counts = comments["body"].str.count(r"\S+")
display(word_counts.describe())
print(f"Comments with fewer than 5 words: {(word_counts < 5).sum():,}")
```

## 4. Explore subreddit activity

The main dataset excludes personality subreddits to reduce the risk of label leakage. The preprocessing report counts comments from those subreddits before the minimum activity filter is applied.

```python
subreddit_counts = comments["subreddit"].value_counts()
print(f"Unique subreddits: {len(subreddit_counts):,}")
print("Top 15 subreddits by comment count")
display(subreddit_counts.head(15))

removed_subreddits = pd.read_csv(REMOVED_SUBREDDITS, index_col=0)
print(f"Comments flagged by the subreddit filter: {removed_subreddits['comment_count'].sum():,}")
display(removed_subreddits)
```

## 5. Summarize the Big Five scores

```python
display(profiles[TRAITS].describe())
```
